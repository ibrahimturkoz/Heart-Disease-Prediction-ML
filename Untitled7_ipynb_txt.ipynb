```python
import cv2
import matplotlib.pyplot as plt
from deepface import DeepFace
import os
import glob
```

```python
def on_islem_adimlarini_goster(img_path):
    img = cv2.imread(img_path)
    if img is None:
        print(f"Resim okunamadı: {img_path}")
        return

    img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

    try:
        objs = DeepFace.extract_faces(
            img_path=img_path,
            target_size=(160, 160),
            detector_backend='opencv',
            enforce_detection=True
        )

        if len(objs) > 0:
            face_img = (objs[0]['face'] * 255).astype('uint8')

            # Aşama Aşamaları Hazırlayalım:
            # 1. Ham Resim (Orijinal)
            # 2. Tespit Edilen & Kesilen Yüz (Crop + Resize 160x160)
            # 3. Kontrast Dengeleme (CLAHE)
            # 4. Keskinleştirilmiş Yüz (Sharpening)

            face_gray = cv2.cvtColor(face_img, cv2.COLOR_RGB2GRAY)
            clahe = cv2.createCLAHE(clipLimit=2.0, tileGridSize=(8, 8))
            face_clahe = clahe.apply(face_gray)

            gaussian = cv2.GaussianBlur(face_img, (0, 0), 2.0)
            face_sharp = cv2.addWeighted(face_img, 1.5, gaussian, -0.5, 0)

            # Ekrana 4 adımı yan yana çizdirme
            fig, axes = plt.subplots(1, 4, figsize=(16, 4))

            axes[0].imshow(img_rgb)
            axes[0].set_title("1. Orijinal Ham Resim")

            axes[1].imshow(face_img)
            axes[1].set_title("2. Kırpılmış & 160x160")

            axes[2].imshow(face_clahe, cmap='gray')
            axes[2].set_title("3. CLAHE (Kontrast)")

            axes[3].imshow(face_sharp)
            axes[3].set_title("4. Keskinleştirilmiş")

            for ax in axes:
                ax.axis('off')

            plt.suptitle(f"İşlenen Dosya: {os.path.basename(img_path)}", fontsize=14)
            plt.tight_layout()
            plt.show()
    except Exception as e:
        print(f"Yüz tespit edilemedi ({img_path}): {e}")
```

```python
# 'benim_yuzum' klasöründeki fotoğrafları tek tek ön işlemden geçirip gösterelim
benim_resimler = glob.glob('dataset/benim_yuzum/*')

print(f"Toplam {len(benim_resimler)} resim için ön işlem adımları gösteriliyor...\n")

for img_path in benim_resimler:
    on_islem_adimlarini_goster(img_path)
```

```python
import cv2
import pickle
import matplotlib.pyplot as plt
from deepface import DeepFace

# Sol paneldeki büyük T harfli dosya adı
test_resim_yolu = "Test.jpg"

try:
    # Modeli yüklüyoruz
    with open('yuz_tanima_modeli.pkl', 'rb') as f:
        clf = pickle.load(f)

                    # enforce_detection=False ile yüz tespitini esnetiyoruz
    objs = DeepFace.represent(img_path=test_resim_yolu, model_name="Facenet", enforce_detection=False)
    test_embedding = objs[0]["embedding"]

                                # Tahmin yapıyoruz
    tahmin = clf.predict([test_embedding])[0]
    olasilik = clf.predict_proba([test_embedding])[0]

                                            # Görselleştirme
    img = cv2.imread(test_resim_yolu)
    img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

    plt.figure(figsize=(6, 6))
    plt.imshow(img_rgb)
    plt.axis('off')

    if tahmin == 1:
        plt.title(f"Tanındı: Kendi Yüzün (%{olasilik[1]*100:.1f})", color='green', fontsize=14)
    else:
        plt.title(f"Tanındı: Başkası (%{olasilik[0]*100:.1f})", color='red', fontsize=14)

    plt.show()
except Exception as e:
        print(f"Hata oluştu: {e}")
```
